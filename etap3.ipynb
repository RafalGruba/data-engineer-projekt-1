```python
from sqlalchemy import create_engine

engine = create_engine('postgresql://postgres:Korsarz7@localhost:5432/postgres')
```

```python
import pandas as pd

df = pd.read_csv('players_data-2025_2026.csv')
df.columns = df.columns.str.lower()
```

```python
df_clean = df.dropna(subset=['age'])
```

```python
df_clean.to_sql('players_python', engine, if_exists='replace', index=False)
```

```python
wynik = pd.read_sql_query("SELECT comp, COUNT(*) FROM players_python GROUP BY comp", engine)
wynik
```

```python
zapytanie = """
WITH avg_by_league AS (
    SELECT comp, AVG(age) AS avg_age
    FROM players_python
    GROUP BY comp
)
SELECT * FROM avg_by_league ORDER BY avg_age DESC
"""

wynik2 = pd.read_sql_query(zapytanie, engine)
wynik2
```
