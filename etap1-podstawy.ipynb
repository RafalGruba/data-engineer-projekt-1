```python
def przywitaj(imie):
    print(f"Cześć, {imie}!")

przywitaj("Rafał")
przywitaj("Ania")

def dodaj(a, b):
    return a + b

wynik = dodaj(3, 5)
print(wynik)   # -> 8

def dodaj_v1(a, b):
    print(a + b)     # tylko WYŚWIETLA wynik

def dodaj_v2(a, b):
    return a + b       # ZWRACA wynik, można go użyć dalej

wynik1 = dodaj_v1(3, 5)   # wypisze "8" na ekranie, ALE wynik1 będzie None!
wynik2 = dodaj_v2(3, 5)   # nic nie wypisze na ekranie, ALE wynik2 będzie równe 8

print(wynik1)   # -> None
print(wynik2)   # -> 8
```

```python
def opisz_gracza(imie, gole=0):
    return f"{imie} strzelił {gole} goli"

print(opisz_gracza("Haaland", 27))
print(opisz_gracza("Mbappe"))   # gole nie podane, użyje domyślnej wartości 0
```

```python
def sklasyfikuj_gole(gole):
    if gole >= 3:
        return "Hat-trick!"
    elif gole >= 1:
        return "Dobry mecz"
    else:
        return "Brak goli"

lista_goli = [0, 2, 3, 1, 4, 0, 5]

for wynik in lista_goli:
    klasyfikacja = sklasyfikuj_gole(wynik)
    print(f"{wynik} goli -> {klasyfikacja}")
```

```python
def oblicz_srednia_goli(lista_goli):
    if not lista_goli: # sprawdzenie, czy lista nie jest pusta
        return 0
    suma = sum(lista_goli)
    srednia = suma / len(lista_goli)
    return srednia

goals = [0, 2, 3, 1, 4, 0, 5]

print(oblicz_srednia_goli(goals))
```

```python
def bezpieczna_konwersja(tekst):
    try:
        liczba = int(tekst)
        print(liczba)
        return liczba
    except ValueError:
        print(f"Nie można przekonwertować '{tekst}' na liczbę całkowitą.")
        return None
bezpieczna_konwersja("25")
bezpieczna_konwersja("dwadziescia piec")
```
